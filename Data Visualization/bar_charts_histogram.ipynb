```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Downloads/diamonds.csv")
df.shape
```

```python
df.head()
```

# Bar Chart

```python
order = ["Fair", "Good", "Very Good", "Premium", "Ideal"]
sns.countplot(data=df, x="cut", color="tab:blue", order=order);
```

## Exercise 2: Bar charts

```python
clarity_counts = df["color"].value_counts(normalize=True) * 100
order = ["D", "E", "F", "G", "H", "I", "J"]
sns.barplot(x=clarity_counts.values, y=clarity_counts.index, order=order);
plt.ylabel("color_grade")
plt.xlabel("proportion");
```

## Exercise 3: Histograms

```python
x_values = df.loc[df["x"] < 11, "x"]

bins = np.arange(0, x_values.max() + 0.25, 0.25)

plt.hist(x_values, bins=bins);
```

## Exercise 4: Histograms

```python
df_tidy = pd.melt(df, value_vars=["x", "y", "z"], var_name = "variable", value_name="values")
```
